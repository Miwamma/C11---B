Todos os exercícios utilizam o dataset **`cidades_brasil.csv`**, que contém informações sobre 25 cidades brasileiras.

```
Índice | Coluna      | Descrição
-------|-------------|----------------------------------
  0    | cidade      | Nome da cidade
  1    | estado      | Sigla do estado
  2    | populacao   | População estimada
  3    | area_km2    | Área em km²
  4    | idh         | Índice de Desenvolvimento Humano
  5    | regiao      | Região do Brasil
  6    | fundacao    | Ano de fundação
```

````python
# Importação e carregamento do dataset
import numpy as np

dados = np.loadtxt('cidades_brasil.csv', delimiter=',', dtype=str, skiprows=1)

# Colunas textuais
cidades    = dados[:, 0]
estados    = dados[:, 1]
regioes    = dados[:, 5]

# Colunas numéricas
populacao  = dados[:, 2].astype(int)
area       = dados[:, 3].astype(int)
idh        = dados[:, 4].astype(float)
fundacao   = dados[:, 6].astype(int)

print(f"Dataset carregado: {len(cidades)} cidades")
````

**a)** Calcule a **densidade demográfica** (população / área) de cada cidade e armazene em um novo array. Exiba as 5 cidades com **maior** densidade (exiba nome e valor).

### Exercício 9 – Manipulação de Texto com `np.char`

**a)** Converta todos os nomes de cidades para **letras maiúsculas** e exiba.

**b)** Verifique quais cidades **começam** com a letra `"S"` e exiba seus nomes. Faça o mesmo para cidades que **começam** com `"B"`.

**c)** Encontre o **índice** (posição) da substring `"do"` em cada nome de cidade. Exiba apenas as cidades onde a substring foi encontrada (índice diferente de -1).

````python
# Resolução do Exercício 9
print(np.char.upper(cidades))

print(cidades[np.char.startswith(cidades, "S")])
print(cidades[np.char.startswith(cidades, "B")])

pos = np.char.find(cidades, "do")

for i in range(len(cidades)):
    if pos[i] != -1:
        print(cidades[i])
````

### Exercício 10 – Persistência de Dados

**a)** Crie uma matriz com as colunas `populacao` e `idh` (use `np.column_stack`). Salve-a em um arquivo de texto **`pop_idh.txt`** usando `np.savetxt` (use `fmt='%.3f'` para manter as casas decimais do IDH).

**b)** Salve o array `fundacao` em um arquivo binário **`fundacao_cidades.npy`** usando `np.save`.

**c)** Carregue ambos os arquivos (`np.loadtxt` e `np.load`) e exiba os dados para confirmar que foram salvos corretamente.

````python
# Resolução do Exercício 10
mat = np.column_stack((populacao, idh))

np.savetxt("pop_idh.txt", mat, fmt="%.3f")

np.save("fundacao_cidades.npy", fundacao)

dados_txt = np.loadtxt("pop_idh.txt")
dados_npy = np.load("fundacao_cidades.npy")

print(dados_txt)
print(dados_npy)
````
