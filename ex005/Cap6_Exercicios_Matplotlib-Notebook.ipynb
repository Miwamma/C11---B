### Exercício 1

Por meio do dataset **paises.csv**, trace dois gráficos de linhas em um mesmo plano cartesiano:
- Um mostrando a **taxa de mortalidade** (`Deathrate`)
- Outro mostrando a **taxa de natalidade** (`Birthrate`)

Ambos devem se referir aos **países da América do Norte**.

> **Dica**: Filtre os países cuja coluna `Region` corresponda à América do Norte. Utilize estilos diferentes para diferenciar as duas linhas e adicione uma legenda ao gráfico.

```python
# Seu código aqui
# Exercício 1
df = pd.read_csv('paises.csv', encoding='utf-8')

df_na = df[df['Region'] == 'NORTHERN AMERICA']

plt.figure()
plt.plot(df_na['ï»¿Country'], df_na['Deathrate'], label='Mortalidade')
plt.plot(df_na['ï»¿Country'], df_na['Birthrate'], label='Natalidade')

plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

### Exercício 2

Por meio do dataset **space.csv**, trace um **gráfico em barras** mostrando quantas **empresas espaciais diferentes** os **EUA** e a **CHINA** possuem.

> **Dica**: Não se esqueça de retirar os resultados repetidos! Utilize o método `.nunique()` ou `.drop_duplicates()` para garantir que cada empresa seja contada apenas uma vez.

```python
# Seu código aqui
# Exercício 2
df = pd.read_csv('space.csv', delimiter=';', encoding='utf-8')

filtro = df[df['Location'].str.contains('USA|China', case=False, na=False)]
empresas = filtro.groupby('Location')['Company Name'].nunique()

empresas.plot(kind='bar')
plt.show()
```

### Exercício 3

Por meio do dataset **space.csv**, trace um **gráfico em torta** ilustrando a **porcentagem de missões** da empresa **Roscosmos** que deram certo e que deram errado.

> **Dica**: Filtre o dataset para obter apenas as missões da Roscosmos e depois conte os valores da coluna de status das missões (`MissionStatus`). Utilize o parâmetro `autopct` para exibir as porcentagens.

```python
# Seu código aqui
# Exercício 3
df = pd.read_csv('space.csv', delimiter=';', encoding='utf-8')

roscosmos = df[df['Company Name'] == 'Roscosmos']
sucesso = (roscosmos['Status Mission'] == 'Success').sum()
falha = (roscosmos['Status Mission'] != 'Success').sum()

plt.figure()
plt.pie([sucesso, falha], labels=['Sucesso', 'Falha'], autopct='%1.1f%%')
plt.show()
```
