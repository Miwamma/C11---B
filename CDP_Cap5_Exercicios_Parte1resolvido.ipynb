## Series

### Questão 1
Crie duas Series com os seguintes valores:
- `seriesAno1`: `{'Java': 16.25, 'C': 16.04, 'Python': 9.85}`
- `seriesAno2`: `{'C': 16.21, 'Python': 12.12, 'Java': 11.68}`

```python
# Seu código aqui
seriesAno1 = pd.Series({'Java': 16.25, 'C': 16.04, 'Python': 9.85})
seriesAno2 = pd.Series({'C': 16.21, 'Python': 12.12, 'Java': 11.68})
```

### Questão 2
Os valores das Series criadas na Questão 1 representam as fatias de mercado (porcentagem) de 3 linguagens de programação populares em dois anos consecutivos. Para cada ano, apresente a **porcentagem total** que elas juntas representam no mercado.

```python
# Seu código aqui
total_ano1 = seriesAno1.sum()
total_ano2 = seriesAno2.sum()

print(total_ano1, total_ano2)
```

### Questão 3
Apresente o **crescimento/declínio** no mercado de cada linguagem do primeiro ano para o segundo ano.

```python
# Seu código aqui
variacao = seriesAno2 - seriesAno1
print(variacao)
```

### Questão 4
Baseado nos resultados da Questão 3, mostre **apenas os dados das linguagens que tiveram crescimento**.

```python
# Seu código aqui
crescimento = variacao[variacao > 0]
print(crescimento)
```

### Questão 5
Se estas porcentagens de crescimento/declínio se mantivessem iguais para os próximos **2 anos**, qual seria a linguagem mais popular?

> **Dica:** use o método `nlargest(1)` no final para retornar rapidamente a label e o maior valor de uma Series.

```python
# Seu código aqui
projecao = seriesAno2 + variacao * 2
print(projecao.nlargest(1))
```

## DataFrame

Para as questões a seguir, utilize o DataFrame exemplo do tópico 5.3 do material. Execute a célula abaixo para criá-lo:

```python
# Criando o DataFrame de referência
np.random.seed(10)
df = pd.DataFrame(
    index=['A', 'B', 'C', 'D', 'E'],
    columns=['W', 'X', 'Y', 'Z'],
    data=np.random.randint(1, 50, [5, 4])
)
print(df)
```

### Questão 6
Calcule a **média dos elementos da coluna X** que são menores que 30.

```python
# Seu código aqui
media_X = df[df['X'] < 30]['X'].mean()
print(media_X)
```

### Questão 7
Apresente a **média dos elementos da linha D** usando a função `loc()` como base e a **soma dos elementos da linha E** usando a função `iloc()` como base.

```python
# Seu código aqui
media_D = df.loc['D'].mean()
soma_E = df.iloc[4].sum()

print(media_D, soma_E)
```

### Questão 8
Faça um *Slicing* na matriz mostrando apenas as **linhas A, C e E** juntamente com as **colunas X e Y**. Em seguida, mostre qual seria a **soma dos elementos** de cada uma destas linhas e cada uma destas colunas.

```python
# Seu código aqui

slice_df = df.loc[['A', 'C', 'E'], ['X', 'Y']]

soma_linhas = slice_df.sum(axis=1)
soma_colunas = slice_df.sum(axis=0)

print(slice_df)
print(soma_linhas)
print(soma_colunas)
```
