## Exercício 1

Crie uma lista preenchida com os 5 primeiros colocados de um Campeonato de Futebol, na ordem de colocação. Depois mostre:

a. Apenas os 3 primeiros colocados;

b. Os últimos 2 colocados;

c. Uma lista com os times em ordem alfabética;

d. Em que posição da tabela se encontra o Barcelona;

```python
# Exercício 1

# Lista com os 5 primeiros colocados
timecolocacao = ['Palmeiras', 'Corinthias', 'São Paulo', 'Barcelona', 'Botafogo']

# a. Apenas os 3 primeiros colocados
print(timecolocacao[:3])

# b. Os últimos 2 colocados
print(timecolocacao[3:])

# c. Uma lista com os times em ordem alfabética
print(sorted(timecolocacao))

# d. Em que posição da tabela se encontra o Barcelona
print(timecolocacao.index('Barcelona'))
```

## Exercício 2

Crie dois conjuntos, um para cada loja. Identifique quais modelos de smartphones cada uma delas vendem. Em seguida, mostre quais modelos no total você terá opção de comprar se visita-las e quais modelos se encontram disponíveis em ambas as lojas;

```python
# Exercício 2

# Conjunto de modelos da loja 1
loja1 = {'iphone', 'motoG', 'Xiamoi', 'sansumg'}

# Conjunto de modelos da loja 2
loja2 = {'iphone', 'Motorola', 'Xiamoi', 'Nokia'}

# Modelos disponíveis no total (união)
total = loja1 | loja2
print(total)

# Modelos disponíveis em ambas as lojas (interseção)
ambas = loja1 & loja2
print(ambas)
```

## Exercício 4

Faça um programa que leia o nome e peso de 3 pessoas e no final mostre o nome da pessoa mais pesada e a mais leve;

```python
# Exercício 4

# Lista para armazenar as pessoas
pessoas = []
peso = []
# Ler dados de 3 pessoas
for n in range(3):
    nome = input("Insira o nome da pessoa: ")
    pessoas.append(nome)
    
    kg = int(input(f"Insira o peso da {n+1}ª pessoa: "))
    peso.append(kg)

# Encontrar a pessoa mais pesada
mais_pesada = pessoas[peso.index(max(peso))]

# Encontrar a pessoa mais leve
mais_leve = pessoas[peso.index(min(peso))]

# Mostrar os resultados
print("Pessoa mais pesada:", mais_pesada)
print("Pessoa mais leve:", mais_leve)
```
