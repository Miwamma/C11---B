### Questão 1 – Controle de Doações em Array

O albergue recebeu doações de alimentos (em kg) durante **10 dias consecutivos**. Use `np.random.seed(42)` e crie um array de **10 números aleatórios inteiros entre 5 e 50** representando os kg doados por dia.

**a)** Exiba as propriedades `size`, `ndim` e `shape` do array;  
**b)** Calcule e exiba a **média**, o **mínimo** e o **máximo** de doações diárias;  
**c)** Mostre o **índice do dia** com maior doação e o **índice do dia** com menor doação;  
**d)** Mostre quantos dias tiveram doação **acima da média**.

````python
# Questão 1
np.random.seed(42)
doacoes = np.random.randint(5, 51, size=10)

print("Array de doações (kg):", doacoes)

# a)
print(doacoes.size, doacoes.ndim, doacoes.shape)

# b)
print(np.mean(doacoes), np.min(doacoes), np.max(doacoes))

# c)
print(np.argmax(doacoes), np.argmin(doacoes))

# d)
print(np.sum(doacoes > np.mean(doacoes)))
````

### Questão 3 – O Mercadinho da Segunda Chance

O Zé das Couves conseguiu um emprego num mercadinho popular e ficou responsável pelo controle de preços. Crie duas **Series do Pandas** com os preços dos produtos em duas semanas diferentes:

- `semana1`: `{'Arroz': 5.50, 'Feijão': 8.90, 'Macarrão': 3.20, 'Sal': 1.80, 'Óleo': 7.40}`
- `semana2`: `{'Feijão': 9.50, 'Óleo': 6.80, 'Arroz': 5.90, 'Sal': 1.80, 'Macarrão': 3.50}`

**a)** Calcule e mostre o **gasto total** na semana 1 e na semana 2;  
**b)** Calcule a **variação de preço** de cada produto (semana2 - semana1);  
**c)** Mostre apenas os produtos que tiveram **aumento de preço**.

````python
# Questão 3
semana1 = pd.Series({'Arroz': 5.50, 'Feijão': 8.90, 'Macarrão': 3.20, 'Sal': 1.80, 'Óleo': 7.40})
semana2 = pd.Series({'Feijão': 9.50, 'Óleo': 6.80, 'Arroz': 5.90, 'Sal': 1.80, 'Macarrão': 3.50})

# a)
print(semana1.sum(), semana2.sum())

# b)
variacao = semana2 - semana1
print(variacao)

# c)
print(variacao[variacao > 0])
````

### Questão 4 – O Grande Censo dos Assistidos

O CRAS fez um censo completo e salvou os dados no arquivo `assistidos.csv`. Carregue o arquivo e responda:

**a)** Exiba **todos os dados** e as **informações gerais** do DataFrame (`.info()`);  
**b)** Calcule a **média de renda** dos assistidos da região `"Norte"`;  
**c)** Mostre o **nome e a renda** do assistido com a **maior renda**;  

````python
# Questão 4 — Carregamento do CSV
df = pd.read_csv("assistidos.csv")

# a)
print(df)
print(df.info())
````

````python
# b)
print(df[df["regiao"] == "Norte"]["renda"].mean())
````

````python
# c) Nome e renda do assistido com maior renda
maior = df.loc[df["renda"].idxmax()]
print(maior["nome"], maior["renda"])
````
