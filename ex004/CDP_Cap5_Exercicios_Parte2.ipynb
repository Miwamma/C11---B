### Questão 2
Encontre o **nome e a região** do país que possui a **maior população** segundo este Dataset.

```python
# Seu código aqui
maior_pop = df.loc[df['Population'].idxmax()]

print(maior_pop['Country'], "-", maior_pop['Region'])
```

### Questão 3
Agrupe os países por **Regiões**. Em seguida, mostre a **média de alfabetização** (`Literacy (%)`) de cada região do planeta.

```python
# Seu código aqui
media_alfabetizacao = df.groupby('Region')['Literacy (%)'].mean()

print(media_alfabetizacao)
```

### Questão 4
Busque o nome de todos os países do Dataset que **não possuem costa marítima** (`Coastline (coast/area ratio) == 0`) e guarde-os em um novo arquivo chamado `noCoast.csv`.

### Questão 5
Faça uma função que receba a **taxa de mortalidade** de cada país (`Deathrate`) e retorne:
- O texto `'Balanced'` caso o valor seja **< 9**;
- O texto `'Urgent'` caso contrário.

Em seguida, crie um campo no Dataset chamado `'Humanitarian Help'` que receba estes valores para cada país. No final, mostre o Dataset para verificar se a inserção da nova coluna foi feita com sucesso.

```python
# Seu código aqui
def classificar_ajuda(valor):
    if valor < 50:
        return 'Urgent'
    elif valor < 100:
        return 'Balanced'
    else:
        return 'Stable'

df['Humanitarian Help'] = df['GDP ($ per capita)'].apply(classificar_ajuda)

print(df[['Country', 'Humanitarian Help']].head())
```
